# Simple notebook to check frequency of differing files

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import os
from tqdm import tqdm
import math as m
import scipy.stats as stats
import scipy.special as sc
from scipy.optimize import curve_fit

import pandas as pd
import sys, os

sys.path.append("/home/nulab4/Documents/MULE_WIP/")

from packs.core import io as io
```

```python
def extract_Hz(file, time):
    # assuming its the standard wd1 stuff
    x = io.load_evt_info(file)
    return x.event_number.nunique() / time
```

# ALPHAS

```python
folder = 'alphas_vac/'
# how long did you let it run? 
time   = 50 # seconds
```

```python
file_names = [f for f in os.listdir(folder) if os.path.isfile(os.path.join(folder, f))]

# sort them cause you're lazy, makes it easier to read also.
# VERY BESPOKE, BE AWARE!
file_names = sorted(file_names, key = lambda s: int(s.split('_')[-1].split('.')[0]))
```

```python
hz = []
for file in file_names:
    hz.append(extract_Hz(f'{folder}{file}', time))
print(file_names)
print(hz)
```
